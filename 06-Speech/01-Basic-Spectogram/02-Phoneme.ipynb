# 02 — Phonemes, G2P, and IPA

## What is a Phoneme?

A **phoneme** is the smallest unit of sound that changes the meaning of a word.

```
"cat"  vs  "bat"  — only the first phoneme differs: /k/ vs /b/
"ship" vs  "chip" — /ʃ/ vs /tʃ/
```

English has **~44 phonemes** but only **26 letters** — spelling does not equal pronunciation:

| Word | Letters | Phonemes |
|------|---------|---------|
| "through" | 7 | 3 (TH R UW) |
| "knight"  | 6 | 3 (N AY T) |
| "fox"     | 3 | 4 (F AA K S) |
| "queue"   | 5 | 2 (K Y UW) |

This mismatch is why TTS/ASR systems almost always convert text to phonemes first — so that **pronunciation is explicit**, not guessed from spelling.

## 1. ARPAbet — The American English Phoneme Set

**ARPAbet** is the most common phoneme alphabet used in American TTS/ASR tools (CMU Pronouncing Dictionary, Festival, Tacotron 2, FastSpeech 2).

It uses ASCII characters so it works in any text file — no special Unicode needed.

| AA | ɑ | f**a**ther, b**o**t | low back |
| AE | æ | b**a**t, c**a**t | low front |
| AH | ʌ | b**u**t, c**u**p | mid central |
| AO | ɔ | b**ou**ght, t**a**lk | mid back |
| AW | aʊ | ab**ou**t, c**ow** | diphthong |
| AY | aɪ | b**i**te, fl**y** | diphthong |
| EH | ɛ | b**e**t, h**e**ad | mid front |
| ER | ɝ | b**ir**d, h**ur**t | r-colored |
| EY | eɪ | b**a**te, d**ay** | diphthong |
| IH | ɪ | b**i**t, h**i**m | high front lax |
| IY | iː | b**ea**t, s**ee** | high front tense |
| OW | oʊ | b**oa**t, g**o** | diphthong |
| OY | ɔɪ | b**oy**, t**oy** | diphthong |
| UH | ʊ | b**oo**k, p**u**t | high back lax |
| UW | uː | b**oo**t, t**oo** | high back tense |

````python
# G2P (Grapheme-to-Phoneme) examples — no install needed
# In production: use g2p-en, phonemizer, or CMUdict

arpabet_examples = {
    "hello":       ["HH", "EH0", "L", "OW1"],
    "world":       ["W",  "ER1", "L", "D"],
    "through":     ["TH", "R",   "UW1"],
    "knight":      ["N",  "AY1", "T"],
    "speech":      ["S",  "P",   "IY1", "CH"],
    "phoneme":     ["F",  "OW1", "N",   "IY0", "M"],
    "alignment":   ["AH0","L",   "AY1", "N",   "M", "AH0", "N", "T"],
    "beautiful":   ["B",  "Y",   "UW1", "T",   "AH0", "F", "AH0", "L"],
    "choir":       ["K",  "W",   "AY1", "ER0"],
    "colonel":     ["K",  "ER1", "N",   "AH0", "L"],
}

print("Grapheme-to-Phoneme (G2P) — English irregularities:")
print(f"  {'Word':<14} {'Phonemes'}")
print("  " + "-" * 48)
for word, phones in arpabet_examples.items():
    stress = [p for p in phones if p[-1].isdigit()]
    print(f"  {word:<14} {' '.join(phones)}")

print()
print("Words with identical pronunciation (homophone pairs):")
homophones = [
    ("through",  ["TH","R","UW1"],  "threw",   ["TH","R","UW1"]),
    ("there",    ["DH","EH1","R"],  "their",   ["DH","EH1","R"]),
    ("flower",   ["F","L","AW1","ER0"], "flour",["F","L","AW1","ER0"]),
    ("knight",   ["N","AY1","T"],   "night",   ["N","AY1","T"]),
]
for w1, p1, w2, p2 in homophones:
    print(f"  {w1:<10} {' '.join(p1)}")
    print(f"  {w2:<10} {' '.join(p2)}  <- same phonemes!")
    print()
````

## 2. IPA — International Phonetic Alphabet

**ARPAbet** is great for English tools, but **IPA** is the global standard — it covers **every sound in every human language** using a single unified symbol system.

| Feature | ARPAbet | IPA |
|---------|---------|-----|
| Coverage | American English only | All human languages |
| Characters | ASCII (AH, EH, ...) | Unicode (ə, ɛ, ʃ, ...) |
| Used in | CMUdict, Festival, Tacotron | Linguistics, multilingual TTS |
| Multilingual TTS | Needs separate system per language | Single phoneme space |

### Why IPA matters for modern TTS/ASR

- **VITS, Bark, CosyVoice**: use IPA as input for multilingual pronunciation control
- **Phonemizer library**: converts text in 100+ languages to IPA via eSpeak
- **Cross-lingual transfer**: IPA lets a model trained on English still produce Thai phonemes

```python
# pip install phonemizer
from phonemizer import phonemize

# English
phonemize("Hello world", backend='espeak', language='en-us', with_stress=True)
# -> 'həlˈoʊ wɜːld'

# Thai
phonemize("สวัสดี", backend='espeak', language='th', with_stress=True)
# -> 'saʔwàt̪diː'
```

````python
# ARPAbet -> IPA mapping for English
arpabet_to_ipa = {
    # Vowels
    "AA": "ɑ",  "AE": "æ",  "AH": "ʌ",  "AO": "ɔ",
    "AW": "aʊ", "AY": "aɪ", "EH": "ɛ",  "ER": "ɝ",
    "EY": "eɪ", "IH": "ɪ",  "IY": "iː", "OW": "oʊ",
    "OY": "ɔɪ", "UH": "ʊ",  "UW": "uː",
    # Consonants
    "B":  "b",  "CH": "tʃ", "D":  "d",  "DH": "ð",
    "F":  "f",  "G":  "g",  "HH": "h",  "JH": "dʒ",
    "K":  "k",  "L":  "l",  "M":  "m",  "N":  "n",
    "NG": "ŋ",  "P":  "p",  "R":  "r",  "S":  "s",
    "SH": "ʃ",  "T":  "t",  "TH": "θ",  "V":  "v",
    "W":  "w",  "Y":  "j",  "Z":  "z",  "ZH": "ʒ",
}

def arpabet_to_ipa_str(phones):
    result = []
    for p in phones:
        base = p.rstrip("012")   # strip stress digit
        stress = p[-1] if p[-1].isdigit() else ""
        ipa_sym = arpabet_to_ipa.get(base, "?")
        if stress == "1": ipa_sym = "ˈ" + ipa_sym   # primary stress
        elif stress == "2": ipa_sym = "ˌ" + ipa_sym  # secondary stress
        result.append(ipa_sym)
    return "/" + "".join(result) + "/"

print("ARPAbet -> IPA conversion:")
print(f"  {'Word':<14} {'ARPAbet':<30} {'IPA'}")
print("  " + "-" * 65)
for word, phones in arpabet_examples.items():
    ipa = arpabet_to_ipa_str(phones)
    print(f"  {word:<14} {' '.join(phones):<30} {ipa}")
````

## 3. Phoneme Inventory by Language

Different languages use different subsets of possible human sounds.
This is why non-native speakers struggle with sounds that don't exist in their language.

````python
# Phoneme inventory comparison
lang_phonemes = {
    "Hawaiian":       {"n": 13,  "vowels": 5,  "consonants": 8,  "note": "fewest phonemes"},
    "Mandarin":       {"n": 21,  "vowels": 9,  "consonants": 21, "note": "tonal language"},
    "Spanish":        {"n": 24,  "vowels": 5,  "consonants": 19, "note": "regular spelling"},
    "English":        {"n": 44,  "vowels": 15, "consonants": 24, "note": "irregular spelling"},
    "Thai":           {"n": 44,  "vowels": 18, "consonants": 21, "note": "tonal + vowel length"},
    "Arabic":         {"n": 28,  "vowels": 6,  "consonants": 28, "note": "pharyngeal sounds"},
    "German":         {"n": 40,  "vowels": 16, "consonants": 24, "note": "umlauts"},
    "!Xóõ (Taa)":    {"n": 141, "vowels": 31, "consonants": 87, "note": "most clicks"},
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart: total phonemes
langs  = list(lang_phonemes.keys())
totals = [v["n"] for v in lang_phonemes.values()]
vowels = [v["vowels"] for v in lang_phonemes.values()]
cons   = [v["consonants"] for v in lang_phonemes.values()]

x = np.arange(len(langs))
axes[0].barh(x, cons,   color="#85C1E9", label="Consonants")
axes[0].barh(x, vowels, left=cons, color="#F9E79F", label="Vowels")
axes[0].set_yticks(x)
axes[0].set_yticklabels(langs, fontsize=10)
axes[0].set_xlabel("Number of phonemes")
axes[0].set_title("Phoneme Inventory by Language")
axes[0].legend(loc="lower right")
for i, (c, v, t) in enumerate(zip(cons, vowels, totals)):
    axes[0].text(t + 1, i, str(t), va="center", fontsize=9)

# Venn-style: shared phonemes between English and Thai
shared_en_th = ["p","b","t","d","k","m","n","ŋ","f","s","h","l","r","w","j",
                "iː","ɛ","æ","ʌ","ɔ","uː"]
only_en = ["ʒ","θ","ð","ʃ","tʃ","dʒ","z","v","g","aɪ","aʊ","eɪ","oʊ","ɔɪ"]
only_th = ["ʔ","pʰ","tʰ","kʰ","tɕ","tɕʰ","ɯ","ɯː","ia","ɯa","ua"]

ax = axes[1]
ax.set_xlim(0, 10); ax.set_ylim(0, 8); ax.axis("off")
ax.set_title("English vs Thai: phoneme overlap", fontsize=11)

circle_en = plt.Circle((3.5, 4), 3.0, color="#AED6F1", alpha=0.5)
circle_th = plt.Circle((6.5, 4), 3.0, color="#ABEBC6", alpha=0.5)
ax.add_patch(circle_en)
ax.add_patch(circle_th)

ax.text(1.5, 7.2, "English only", ha="center", fontsize=9, color="#2471A3")
ax.text(8.5, 7.2, "Thai only", ha="center", fontsize=9, color="#1E8449")
ax.text(5.0, 7.2, "Shared", ha="center", fontsize=9, color="#555")

for i, ph in enumerate(only_en[:6]):
    ax.text(2.0, 6.0 - i*0.7, f"/{ph}/", ha="center", fontsize=8, color="#1A5276")
ax.text(2.0, 1.8, f"+{len(only_en)-6} more", ha="center", fontsize=8, color="#1A5276")

for i, ph in enumerate(shared_en_th[:7]):
    ax.text(5.0, 6.2 - i*0.65, f"/{ph}/", ha="center", fontsize=8, color="#333")
ax.text(5.0, 1.5, f"+{len(shared_en_th)-7} more", ha="center", fontsize=8, color="#333")

for i, ph in enumerate(only_th[:5]):
    ax.text(8.0, 6.0 - i*0.7, f"/{ph}/", ha="center", fontsize=8, color="#1A5276")
ax.text(8.0, 2.5, f"+{len(only_th)-5} more", ha="center", fontsize=8, color="#145A32")

plt.tight_layout()
plt.savefig("figures/phoneme_inventory.png", dpi=110, bbox_inches="tight")
plt.show()
print("Saved figures/phoneme_inventory.png")
````

## 4. The G2P Pipeline in TTS Systems

The full text preprocessing pipeline for a modern TTS system:

```
Raw text:  "Dr. Smith bought 3 books."
              |
    Text Normalization
    (expand abbreviations, numbers)
              |
         "Doctor Smith bought three books."
              |
    G2P Conversion
    (text -> phoneme sequence)
              |
    D AH1 K T ER0  S M IH1 TH  B AO1 T  TH R IY1  B UH1 K S
              |
    Optional: add stress / prosody labels
              |
    Acoustic Model (Tacotron2 / FastSpeech2 / VITS)
              |
    Mel spectrogram -> Vocoder -> Waveform
```

**Why G2P instead of raw characters?**
- "read" (present) vs "read" (past) — same spelling, different pronunciation
- Numbers: "2024" → "twenty twenty four" (context-dependent)
- Abbreviations: "Dr." → "Doctor" vs "Drive" (context-dependent)
- Better model generalization: phonemes are language-level units, characters are writing-level

````python
# Demonstrate text normalization + G2P pipeline
text_norm_examples = [
    ("Dr. Smith",       "Doctor Smith"),
    ("3 cats",          "three cats"),
    ("$50",             "fifty dollars"),
    ("AI",              "A I"),
    ("2024",            "twenty twenty four"),
    ("e.g.",            "for example"),
    ("10:30am",         "ten thirty A M"),
    ("www.example.com", "www dot example dot com"),
]

# Manual G2P for a few normalized words
g2p_manual = {
    "doctor":  ["D", "AH1", "K", "T", "ER0"],
    "smith":   ["S", "M", "IH1", "TH"],
    "three":   ["TH", "R", "IY1"],
    "cats":    ["K", "AE1", "T", "S"],
    "fifty":   ["F", "IH1", "F", "T", "IY0"],
    "dollars": ["D", "AA1", "L", "ER0", "Z"],
}

print("Text Normalization:")
print(f"  {'Raw text':<25} -> {'Normalized'}")
print("  " + "-" * 50)
for raw, norm in text_norm_examples:
    print(f"  {raw:<25} -> {norm}")

print()
print("G2P for normalized words:")
print(f"  {'Word':<12} {'ARPAbet':<30} {'IPA'}")
print("  " + "-" * 58)
for word, phones in g2p_manual.items():
    ipa = arpabet_to_ipa_str(phones)
    print(f"  {word:<12} {' '.join(phones):<30} {ipa}")
````

## Summary

| Concept | Key points |
|---------|-----------|
| **Phoneme** | Smallest meaning-changing sound unit; ~44 in English |
| **ARPAbet** | ASCII phoneme set for American English; used in CMUdict, Tacotron, FastSpeech2 |
| **IPA** | Universal phoneme alphabet; covers 100+ languages; used in multilingual TTS |
| **G2P** | Converts spelling to phonemes; handles irregularities like "knight", "through" |
| **Text normalization** | Expands abbreviations/numbers before G2P |
| **Stress markers** | ARPAbet 0/1/2 = unstressed/primary/secondary; affects pitch and duration |
